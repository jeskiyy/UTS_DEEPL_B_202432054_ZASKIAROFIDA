<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432054_ZaskiaRofida
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Menetapkan seed agar hasil eksperimen lebih konsisten
torch.manual_seed(42)
np.random.seed(42)

# Menentukan perangkat komputasi
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Perangkat yang digunakan:", device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432054_ZaskiaRofida
# 1. Membaca dataset
df = pd.read_csv("london_merged.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])
df = df.sort_values("timestamp").reset_index(drop=True)

print("Ukuran dataset:", df.shape)
print("Periode data:", df["timestamp"].min(), "sampai", df["timestamp"].max())
print("\nLima data pertama:")
display(df.head())
print("\nJumlah data kosong per kolom:")
display(df.isnull().sum())

# 2. Mengambil target yang akan diprediksi
# Model menggunakan 24 catatan sebelumnya untuk memprediksi 1 jam berikutnya.
target = df["cnt"].astype("float32").values

# Membagi data secara berurutan: 80% awal untuk training dan 20% akhir untuk testing.
# Pembagian kronologis dilakukan agar urutan waktu tidak tercampur.
split_idx = int(len(target) * 0.8)
train_values = target[:split_idx]

# Normalisasi menggunakan nilai minimum dan maksimum dari data training saja.
train_min = train_values.min()
train_max = train_values.max()
target_scaled = (target - train_min) / (train_max - train_min)

sequence_length = 24

def create_train_sequences(data, split_point, seq_length):
    X, y = [], []
    for i in range(seq_length, split_point):
        X.append(data[i-seq_length:i])
        y.append(data[i])
    return np.array(X), np.array(y)

def create_test_sequences(data, split_point, seq_length):
    X, y = [], []
    for i in range(split_point, len(data)):
        X.append(data[i-seq_length:i])
        y.append(data[i])
    return np.array(X), np.array(y)

X_train, y_train = create_train_sequences(target_scaled, split_idx, sequence_length)
X_test, y_test = create_test_sequences(target_scaled, split_idx, sequence_length)

# Menambahkan dimensi fitur agar bentuk input sesuai dengan LSTM: (sampel, urutan, fitur).
X_train = torch.tensor(X_train, dtype=torch.float32).unsqueeze(-1)
y_train = torch.tensor(y_train, dtype=torch.float32).unsqueeze(-1)
X_test = torch.tensor(X_test, dtype=torch.float32).unsqueeze(-1)
y_test = torch.tensor(y_test, dtype=torch.float32).unsqueeze(-1)

print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))
print("Bentuk X_train:", X_train.shape)
print("Bentuk y_train:", y_train.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432054_ZaskiaRofida
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=1):
        super(SequenceModel, self).__init__()
        # Layer LSTM menerima urutan 24 catatan data dengan satu fitur, yaitu cnt.
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )
        # Layer Linear mengubah keluaran LSTM menjadi satu nilai prediksi.
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        # Mengambil keluaran LSTM pada seluruh urutan.
        lstm_out, (hidden, cell) = self.lstm(x)
        # Menggunakan keluaran pada catatan terakhir dalam urutan input.
        last_output = lstm_out[:, -1, :]
        prediction = self.fc(last_output)
        return prediction
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432054_ZaskiaRofida
# 1. Menentukan hyperparameter
epochs = 30
learning_rate = 0.001
batch_size = 64
hidden_size = 64

# 2. Membuat DataLoader untuk data training
train_dataset = TensorDataset(X_train, y_train)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=False)

# Inisialisasi model, loss function, dan optimizer
model = SequenceModel(input_size=1, hidden_size=hidden_size).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Menyimpan nilai loss setiap epoch
train_losses = []

# 4. Proses training
for epoch in range(epochs):
    model.train()
    total_loss = 0.0

    for batch_X, batch_y in train_loader:
        batch_X = batch_X.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * batch_X.size(0)

    average_loss = total_loss / len(train_loader.dataset)
    train_losses.append(average_loss)

    if (epoch + 1) % 5 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {average_loss:.6f}")

print("Training selesai.")
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432054_ZaskiaRofida
# 1. Grafik loss selama training
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, linewidth=2, label="Training Loss")
plt.title("Grafik Penurunan Loss Selama Pelatihan")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

# 2. Menghasilkan prediksi pada data testing
model.eval()
with torch.no_grad():
    predictions_scaled = model(X_test.to(device)).cpu().numpy()

# Mengembalikan skala data ke jumlah penyewaan sebenarnya
predictions = predictions_scaled.flatten() * (train_max - train_min) + train_min
actual = y_test.numpy().flatten() * (train_max - train_min) + train_min

# Menghitung metrik evaluasi
mae = np.mean(np.abs(actual - predictions))
rmse = np.sqrt(np.mean((actual - predictions) ** 2))
print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")

# 3. Grafik perbandingan aktual dan prediksi
plt.figure(figsize=(14, 5))
plt.plot(actual[:300], label="Data Aktual", linewidth=1.5)
plt.plot(predictions[:300], label="Hasil Prediksi", linewidth=1.5)
plt.title("Perbandingan Jumlah Penyewaan Aktual dan Prediksi (300 Jam Pertama Data Uji)")
plt.xlabel("Urutan Jam pada Data Uji")
plt.ylabel("Jumlah Penyewaan Sepeda")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```
